## 2.
What is the best wine I can buy for a given amount of money? Create a `Series` whose index is wine prices and whose values is the maximum number of points a wine costing that much was given in a review. Sort the values by price, ascending (so that `4.0` dollars is at the top and `3300.0` dollars is at the bottom).

```python
#q2.hint()
#q2.solution()
reviews.groupby('country').count().head(5)
```

```python
# Here is one interesting stuff :-
# you can use value_counts() similar fucntionality using groupby()
# normal value_counts() approach
countries = reviews.country.value_counts().sort_index().sort_values(ascending = False)

# here is groupby() appraoch
c2 = reviews.groupby('country').count().description.sort_index().sort_values(ascending =False)

# you should observe that we first sorted in both index and then values (you can do 
# vice versa but order should be same). if you don't sort, the Index won't align
# ---
# if you check the count, 'china' and 'serbia', they have same count but different positon in series
# therefore sorting is need to get same index position in series
c2 ==  countries
```
