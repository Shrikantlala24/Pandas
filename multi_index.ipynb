```python
import numpy as np
import pandas as pd
```

```python
index_col = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]\

MI = pd.MultiIndex.from_tuples(index_col)

S1 = pd.Series([1,2,3,4,5,6,7,8], index=MI)
S1
```

```python
# let's create it from cartesian product
MI2 = pd.MultiIndex.from_product([['cse','ece'],[2019,2020,2021,2022]])
print(MI2)
S2 = pd.Series([1,2,3,4,5,6,7,8],index = MI2)
S2
```
