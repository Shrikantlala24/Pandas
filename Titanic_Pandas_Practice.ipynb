## Dataset Information

Source: `seaborn-data` (public dataset)

The Titanic dataset contains passenger data from the RMS Titanic with survival information. This dataset is excellent for practicing **data cleaning**, **handling missing values**, and **exploratory data analysis** on real-world messy data.

### Columns

- **survived** – Whether the passenger survived (0 = No, 1 = Yes)
- **pclass** – Ticket class (1 = 1st class, 2 = 2nd class, 3 = 3rd class)
- **sex** – Gender (male / female)
- **age** – Age in years (contains missing values)
- **sibsp** – Number of siblings/spouses aboard
- **parch** – Number of parents/children aboard
- **fare** – Ticket fare paid
- **embarked** – Port of embarkation (C = Cherbourg, Q = Queenstown, S = Southampton)
- **class** – Ticket class (name) – First, Second, Third
- **who** – Adult man, adult woman, or child
- **adult_male** – Boolean indicating if passenger is an adult male
- **deck** – Deck letter (contains missing values)
- **embark_town** – Full port name
- **alive** – Whether survived (yes / no)
- **alone** – Boolean indicating if traveling alone
