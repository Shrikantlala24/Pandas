## Level 1: Basic Concatenation

### Question 1: Vertical Concatenation
**Task**: Concatenate `customers_orders` and `additional_orders` DataFrames vertically to create a complete order history. Reset the index after concatenation.

**Concepts**: `pd.concat()`, `axis=0`, `ignore_index=True`

**Expected Output**: A DataFrame with 10 rows (7 + 3 orders)

```python
orders = pd.concat([customers_orders, additional_orders], ignore_index=True)
orders
```

```python
# this is the way where we sustain the indexes of both datasets and conatenate them

orders_with_keys = pd.concat([customers_orders, additional_orders], keys = ['original', 'additional'])
orders_with_keys
# to access elements from both datasets after concatenation with keys
# we can use .loc method

# # ! method 1
# orders_with_keys.loc['original'].iloc[0]

# # ! method 2
# orders_with_keys.loc[('original', 0)]
```

## Level 2: Inner and Outer Joins

### Question 3: Inner Join
**Task**: Merge `customers_orders` with `order_details` on `order_id` using an inner join. Count how many records are in the result and explain why.

**Concepts**: `pd.merge()`, `how='inner'`, `on='column_name'`

**Expected Output**: Should have fewer rows than both original DataFrames

```python
customers_orders.merge(order_details,on = 'order_id')

# in this kind of join only those rows which are present in both datasets will be shown
# more precisely the order_id should be present on both in order to map the combined rows
```

### Question 4: Outer Join (Full Join)
**Task**: Perform a full outer join between `customers_orders` and `order_details` on `order_id`. Identify which orders exist in one DataFrame but not the other.

**Concepts**: `pd.merge()`, `how='outer'`

**Expected Output**: All orders from both DataFrames, with NaN values where data is missing

```python
customers_orders.merge(order_details,on = 'order_id', how = 'outer')
# sab khichdi aa jayegi
# type(np.nan)
```

### Question 5: Left Join
**Task**: Perform a left join with `customers_orders` as the left DataFrame and `order_details` as the right DataFrame. What happens to order_id 1008 and 1009?

**Concepts**: `pd.merge()`, `how='left'`

**Expected Output**: All customer orders preserved, NaN for orders without product details

```python
customers_orders.merge(order_details,on = 'order_id', how = 'left')

# orders whose customer dtails are present but product details are missing
```

### Question 6: Right Join
**Task**: Perform a right join with `customers_orders` as the left DataFrame and `order_details` as the right DataFrame. Compare the results with the left join from Question 5.

**Concepts**: `pd.merge()`, `how='right'`

**Expected Output**: All order details preserved, NaN for orders without customer information

```python
customers_orders.merge(order_details,on = 'order_id', how = 'right')

# orders whose products details are present but customer details are missing
```

## Level 3: Multi-Key Merges and Complex Scenarios

### Question 7: Three-Way Merge
**Task**: Create a comprehensive dataset by merging all three DataFrames:
1. First merge `customers_orders` with `order_details`
2. Then merge the result with `product_inventory`

Use appropriate join types to preserve all order information.

**Concepts**: Sequential merging, choosing appropriate join types

**Expected Output**: A comprehensive view of orders with customer, product, and inventory details

```python
main_inner = customers_orders.merge(order_details, on = 'order_id',).merge(product_inventory, on = 'product_id')
main_inner

#  this is a proper dataset without any missing values which could be created when merging all three datasets
```

```python
main_outer = customers_orders.merge(order_details, on = 'order_id', how='outer').merge(product_inventory, on = 'product_id', how = 'outer')
main_outer


# one hell of a dataset with so many missing values due to outer joins
```
