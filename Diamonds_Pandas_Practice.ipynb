### Q27. Categorize diamonds based on price

Tasks:
- Create a new column `price_category`:
  - `budget` : price in [0, 1000)
  - `mid-range` : [1000, 5000)
  - `premium` : [5000, 10000)
  - `luxury` : [10000, ∞)
- Count how many diamonds fall in each category.


```python

# ! i don't know this method yet but the work is same as we do in other approach
# diamonds['Price Category'] = pd.cut(diamonds['price'], bins=[0,500,1500,3000,10000], labels=['Low','Medium','High','Very High'])
# diamonds['Price Category']

def categorize(col):
    if col < 1000:
        return 'Budget'
    elif 1000 <= col < 5000:
        return 'Mid-Range'
    elif 5000 <= col < 10000:
        return 'Premium'
    else:
        return 'Luxury'

diamonds['Price_Category'] = diamonds['price'].apply(categorize)
```
